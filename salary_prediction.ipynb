# Salary Prediction using Simple Linear Regression

This project predicts an employee's salary based on years of experience using Simple Linear Regression.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

## 1. Load the Dataset

```python
df = pd.read_csv("Salary.csv")
print("Dataset shape:", df.shape)
df.head()
```

## 2. Explore the Dataset

```python
df
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

## 3. Define the Feature and Target

```python
# Independent variable
X = df[["YearsExperience"]]

# Dependent variable
y = df["Salary"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFirst 5 values of X:")
print(X.head())

print("\nFirst 5 values of y:")
print(y.head())
```

## 4. Create and Train the Linear Regression Model

```python
regr = LinearRegression()
regr.fit(X, y)
```

```python
a = regr.coef_[0]
b = regr.intercept_

print("Coefficient (a):", a)
print("Intercept (b):", b)
print("\nLinear Regression equation: y = ax + b")
```

## 5. Make Predictions

```python
y_pred = regr.predict(X)
print("First 5 predictions:")
print(y_pred[:5])
```

## 6. Visualize the Data and Regression Line

```python
plt.figure(figsize=(8, 5))
plt.scatter(X["YearsExperience"], y, label="Actual Data")
plt.plot(X["YearsExperience"], y_pred, label="Regression Line")
plt.xlabel("Years Experience")
plt.ylabel("Salary")
plt.title("Salary vs Years Experience")
plt.legend()
plt.grid(True)
plt.show()
```

## 7. Predict Salary for 4 Years of Experience

```python
prediction = regr.predict([[4]])
print("Predicted salary for 4 years of experience:", prediction[0])
```

## 8. Model Evaluation

```python
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Squared Error (MSE):", mse)
print("R² Score:", r2)
```

## 9. Manual MSE Calculation

```python
mse_manual = np.mean((y.values - y_pred) ** 2)
print("Manual MSE:", mse_manual)
print("MSE values are equal:", np.isclose(mse, mse_manual))
```

## 10. Manual R² Calculation

```python
ss_res = np.sum((y.values - y_pred) ** 2)
ss_tot = np.sum((y.values - y.mean()) ** 2)
r2_manual = 1 - (ss_res / ss_tot)

print("Manual R² Score:", r2_manual)
print("R² values are equal:", np.isclose(r2, r2_manual))
```

## 11. Manual Prediction Using y = ax + b

```python
y_pred_manual = a * X["YearsExperience"] + b
print(y_pred_manual.head())

print("Predictions are approximately equal:", np.allclose(y_pred, y_pred_manual.values))
```

## 12. Variance Calculation

```python
variance_manual = np.sum((y - y.mean()) ** 2) / len(y)
print("Manual Variance:", variance_manual)
print("NumPy Variance:", np.var(y))
print("Variance values are equal:", np.isclose(variance_manual, np.var(y)))
```

## 13. Final Results

```python
print("========== Final Results ==========")
print("Coefficient:", a)
print("Intercept:", b)
print("MSE:", mse)
print("R² Score:", r2)
print("===================================")
```

## Conclusion

This project demonstrates how Simple Linear Regression can be used to predict employee salary based on years of experience. The model is evaluated using Mean Squared Error (MSE) and R² Score, with manual calculations included to understand the underlying mathematics.
